# TSCGlue Classification Demo

Trains `TSCGlueEnhancedV2` and then `TSCGlueEnhancedV3` on a small UCR dataset. The two stacks
are identical except that V3 computes `hydra` on the GPU, so both run with `n_gpus=1` — mantis
and chronos2 are on the GPU either way, which leaves hydra's device as the only difference.

```python
import tempfile

import numpy as np
from sklearn.metrics import accuracy_score

from tscglue import utils
from tscglue.models import TSCGlueEnhancedV4
```

```python
DATASET = "SyntheticControl"
```

```python
X_train, y_train, X_test, y_test = utils.load_dataset(DATASET)
print(f"Train: {X_train.shape}, Test: {X_test.shape}")
print(f"Classes: {np.unique(y_train)}")
```

```python
with tempfile.TemporaryDirectory() as tmp_dir:
    v3 = TSCGlueEnhancedV4(
        random_state=0,
        n_jobs=4,
        n_gpus=1,
        verbose=1,
        eval_metric="log_loss",
        preset="medium",
        runs_dir=tmp_dir,
    )
    v3.fit(X_train, y_train)
    print(f"V3 test accuracy: {accuracy_score(y_test, v3.predict(X_test)):.4f}")
```
